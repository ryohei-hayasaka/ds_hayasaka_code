# Autograd

```python
import torch
```

```python
x = torch.tensor(2., requires_grad=True) # intではなくfloat32にするため2.とする
y = torch.tensor(3., requires_grad=True)

z = y * torch.log(x) + torch.sin(y)
z.backward() # 勾配計算
```

```python
x.grad
```

```python
y.grad
```

```python
z = y * torch.log(x) + torch.sin(y)
z.backward()
```

```python
x.grad # 勾配は累積される　1.5+1.5=3.0になっている
```

### 中間ノードの勾配

```python
x = torch.tensor(2., requires_grad=True) # intではなくfloat32にするため2.とする
y = torch.tensor(3., requires_grad=True)

sum_xy = x + y
sum_xy.retain_grad()

z = sum_xy**2
z.backward() 
```

```python
print(sum_xy.grad)
print(x.grad)
print(y.grad)
```
