## Tensorの操作

```python
tensor_example = torch.rand((2, 3, 4))
np_example = np.random.rand(2, 3, 4)
print(tensor_example, np_example)
```

```python
tensor_example.shape
```

```python
np_example.shape
```

```python
# 転置
permuted_tensor = torch.permute(tensor_example, (1, 0, 2))
transposed_np = np.transpose(np_example, (1, 0, 2))
print(permuted_tensor.shape)
print(transposed_np.shape)

transposed_tensor = torch.transpose(tensor_example, 0, 1)
print(transposed_tensor.shape)

# 2次元（rankが2つのtensor）の場合は.Tや.t()で転置することができる
```

```python
# reshape
reshaped_tensor = torch.reshape(tensor_example, (6, 4))
reshaped_np = np.reshape(np_example, (6, 4))
print(reshaped_tensor.shape)
print(reshaped_np.shape)
```

```python
reshaped_tensor[0] = 0
reshaped_tensor
```

```python
tensor_example
```

#### ↑reshapeは基本的にはコピーではなく、同じメモリを使用し元々の変数にアクセスするため、reshape後の変数を変更すると元々の変数も変更される

```python
# メモリが連続でなければreshapeはコピーを返す
x = torch.tensor([[1, 2], [3, 4], [5, 6]])
print(x)
print(x.T)
y = x.T
```

```python
# 連続なメモリに保存されていない
y.is_contiguous() #連続→Trueを返す
```

```python
x.is_contiguous()
```

```python
z = y.reshape(-1)
```

```python
z[0] = 0
z
```

```python
y
```

#### ↑zは連続でないのでreshapeでコピーを返しているため、yには影響がない

```python
z = y.view(-1)
```

#### ↑.view()は連続の場合のみreshapeし、連続でない場合はエラーを返すのでtensorを無駄に増やすことを防ぐことができる

```python
# flatten
flattened_tensor = torch.flatten(tensor_example)
flattened_np = np_example.flatten
print(flattened_tensor.shape)
print(flattened_np.shape)
```

```python
# suqueeze
tensor_example = torch.tensor([[[1], [2], [3]]])
np_example = np.array([[[1], [2], [3]]])
print(tensor_example.shape)
print(np_example.shape)
```

```python
squeezed_tensor = torch.squeeze(tensor_example)
squeezed_np = np.squeeze(np_example)
print(squeezed_tensor.shape)
print(squeezed_np.shape)
```

```python
unsqueezed_tensor = torch.unsqueeze(tensor_example, 0)
expand_dims_np = np.expand_dims(np_example, 0)
print(unsqueezed_tensor.shape)
print(expand_dims_np.shape)
```

```python
tensor_example[None, :, :, :].shape
```

#### ↑この書き方でもランクを増加させることができる
